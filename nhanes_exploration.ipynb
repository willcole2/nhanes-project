```python
# Load the Weight History dataset
whq = pd.read_sas("data/WHQ_L.xpt")

# Check dimensions and variable names
print("Number of rows and columns:", whq.shape)
print(whq.columns.tolist())
```

```python
# Descriptive statistics for Weight History
whq.describe()
```

```python
import numpy as np

# Make a copy of the original dataset
whq_clean = whq.copy()

# Replace special missing-value codes
whq_clean["WHD010"] = whq_clean["WHD010"].replace(
    [7777, 9999], np.nan
)

whq_clean["WHD020"] = whq_clean["WHD020"].replace(
    [7777, 9999], np.nan
)

whq_clean["WHD050"] = whq_clean["WHD050"].replace(
    [7777, 9999], np.nan
)

whq_clean["WHQ070"] = whq_clean["WHQ070"].replace(
    [7, 9], np.nan
)

# Examine descriptive statistics after cleaning
whq_clean.describe()
```

```python
# Load the Physical Activity dataset
paq = pd.read_sas("data/PAQ_L.xpt")

# Check dimensions and variable names
print("Number of rows and columns:", paq.shape)
print(paq.columns.tolist())
```

```python
# Descriptive statistics for Physical Activity
paq.describe()
```

```python
# Examine the response frequencies for moderate and vigorous activity

print("Moderate activity frequency:")
print(paq["PAD790Q"].value_counts(dropna=False).sort_index())

print("\nVigorous activity frequency:")
print(paq["PAD810Q"].value_counts(dropna=False).sort_index())
```

```python
# Examine the units used to report physical activity frequency

print("Moderate activity units:")
print(paq["PAD790U"].value_counts(dropna=False))

print("\nVigorous activity units:")
print(paq["PAD810U"].value_counts(dropna=False))
```

```python
# Examine how activity frequency corresponds to its reporting unit

print("Moderate activity:")
display(pd.crosstab(
    paq["PAD790Q"],
    paq["PAD790U"],
    dropna=False
))

print("Vigorous activity:")
display(pd.crosstab(
    paq["PAD810Q"],
    paq["PAD810U"],
    dropna=False
))
```

```python
# Check whether near-zero activity frequencies correspond to missing durations

print("Moderate activity:")
print(
    paq.loc[paq["PAD790Q"].between(0, 1e-50), "PAD800"]
    .value_counts(dropna=False)
)

print("\nVigorous activity:")
print(
    paq.loc[paq["PAD810Q"].between(0, 1e-50), "PAD820"]
    .value_counts(dropna=False)
)
```

```python
import numpy as np

# Preserve the original dataset
paq_clean = paq.copy()

# Clean moderate and vigorous activity frequencies
for col in ["PAD790Q", "PAD810Q"]:
    paq_clean[col] = paq_clean[col].replace(
        [7777, 9999], np.nan
    )

    # Correct near-zero SAS values to actual zero
    paq_clean.loc[
        paq_clean[col].between(0, 1e-50), col
    ] = 0

# Verify the cleaning
for col in ["PAD790Q", "PAD810Q"]:
    print(f"\n{col}:")
    print("Zero activity:", (paq_clean[col] == 0).sum())
    print("Missing:", paq_clean[col].isna().sum())
```

```python
# Examine special response codes for activity duration

for col in ["PAD800", "PAD820"]:
    print(f"\n{col}:")
    print("Refused (7777):", (paq_clean[col] == 7777).sum())
    print("Don't know (9999):", (paq_clean[col] == 9999).sum())
    print("Missing:", paq_clean[col].isna().sum())
    print("Minimum:", paq_clean[col].min())
    print("Maximum:", paq_clean[col].max())
```

```python
# Clean physical activity duration variables

for col in ["PAD800", "PAD820"]:
    paq_clean[col] = paq_clean[col].replace(
        [7777, 9999], np.nan
    )

# Check the results
paq_clean[["PAD800", "PAD820"]].describe()
```

```python
# Convert frequency unit codes from bytes to strings
for col in ["PAD790U", "PAD810U"]:
    paq_clean[col] = paq_clean[col].str.decode("utf-8")

# Convert reported activity frequencies to weekly equivalents
def sessions_per_week(frequency, unit):
    if pd.isna(frequency):
        return np.nan
    if frequency == 0:
        return 0
    if unit == "D":
        return frequency * 7
    if unit == "W":
        return frequency
    if unit == "M":
        return frequency * 12 / 52
    if unit == "Y":
        return frequency / 52
    return np.nan

paq_clean["moderate_sessions_week"] = paq_clean.apply(
    lambda row: sessions_per_week(row["PAD790Q"], row["PAD790U"]),
    axis=1
)

paq_clean["vigorous_sessions_week"] = paq_clean.apply(
    lambda row: sessions_per_week(row["PAD810Q"], row["PAD810U"]),
    axis=1
)

# Examine the new variables
paq_clean[
    ["moderate_sessions_week", "vigorous_sessions_week"]
].describe()
```

```python
# Examine participants reporting more than 14 activity occasions per week

print("Unusually high moderate activity:")
display(
    paq_clean.loc[
        paq_clean["moderate_sessions_week"] > 14,
        ["SEQN", "PAD790Q", "PAD790U",
         "PAD800", "moderate_sessions_week"]
    ].sort_values("moderate_sessions_week", ascending=False)
)

print("Unusually high vigorous activity:")
display(
    paq_clean.loc[
        paq_clean["vigorous_sessions_week"] > 14,
        ["SEQN", "PAD810Q", "PAD810U",
         "PAD820", "vigorous_sessions_week"]
    ].sort_values("vigorous_sessions_week", ascending=False)
)
```

```python
# Calculate estimated minutes of activity per week

paq_clean["moderate_min_week"] = (
    paq_clean["moderate_sessions_week"] * paq_clean["PAD800"]
)

paq_clean["vigorous_min_week"] = (
    paq_clean["vigorous_sessions_week"] * paq_clean["PAD820"]
)

# Participants reporting no activity should have zero minutes
paq_clean.loc[
    paq_clean["moderate_sessions_week"] == 0,
    "moderate_min_week"
] = 0

paq_clean.loc[
    paq_clean["vigorous_sessions_week"] == 0,
    "vigorous_min_week"
] = 0

# Examine the resulting distributions
paq_clean[
    ["moderate_min_week", "vigorous_min_week"]
].describe()
```

```python
# Calculate total leisure-time physical activity minutes per week

paq_clean["total_ltpa_min_week"] = paq_clean[
    ["moderate_min_week", "vigorous_min_week"]
].sum(axis=1, min_count=2)

# Examine the new variable
paq_clean["total_ltpa_min_week"].describe()
```

```python
# Merge Demographics, Weight History, and Physical Activity

nhanes = (
    demo
    .merge(whq_clean, on="SEQN", how="left", validate="one_to_one")
    .merge(paq_clean, on="SEQN", how="left", validate="one_to_one")
)

# Restrict to adults aged 20 or older
nhanes_adults = nhanes[nhanes["RIDAGEYR"] >= 20].copy()

# Examine the merged dataset
print("Merged dataset:", nhanes.shape)
print("Adult sample:", nhanes_adults.shape)
```

```python
# Examine missing data for our primary research variables

variables = [
    "WHD020",             # Current weight
    "WHD050",             # Weight one year ago
    "WHQ070",             # Weight-loss attempts
    "moderate_min_week",  # Moderate physical activity
    "vigorous_min_week",  # Vigorous physical activity
    "total_ltpa_min_week" # Total physical activity
]

missing_summary = pd.DataFrame({
    "Valid N": nhanes_adults[variables].notna().sum(),
    "Missing N": nhanes_adults[variables].isna().sum(),
    "Missing %": nhanes_adults[variables].isna().mean() * 100
})

missing_summary.round(2)
```

```python
# Calculate percentage weight change

nhanes_adults["pct_weight_change"] = (
    (nhanes_adults["WHD020"] - nhanes_adults["WHD050"])
    / nhanes_adults["WHD050"]
) * 100

# Examine descriptive statistics
nhanes_adults["pct_weight_change"].describe()
```

```python
# Identify participants with complete data for the primary analysis

primary_vars = [
    "pct_weight_change",
    "WHQ070",
    "total_ltpa_min_week",
    "RIDAGEYR",
    "RIAGENDR"
]

primary_sample = nhanes_adults.dropna(
    subset=primary_vars
).copy()

print("Total adults:", len(nhanes_adults))
print("Complete cases:", len(primary_sample))
print(
    "Percentage retained:",
    round(len(primary_sample) / len(nhanes_adults) * 100, 2)
)
```
